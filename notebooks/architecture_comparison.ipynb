## Architectural confound and conclusions

The preserved v1 encoder-decoder uses Tanh output, while v2 uses ReLU after its
final convolution and batch normalization. They differ in output activation as
well as skip connections: v2 cannot output negative normalized a*/b* values.
This comparison cannot isolate the effect of skip connections. Changing the
activation requires a separately named architecture and a new experiment.
No quality ranking, SOTA claim, or improvement in diversity is established.
Curate the same examples, including failures, only after obtaining new outputs
and recording checkpoint, dataset, file, and preprocessing provenance.
